# Домашнее задание: Bash-скрипты

**Формат сдачи:** пять отдельных скриптов `.sh` (не код в ячейках ноутбука). Каждый должен запускаться командой `bash имя_файла.sh`.

### Имена файлов

Используйте шаблон **`фамилия_XX.sh`**, где фамилия записана строчными буквами, а `XX` — двузначный номер задачи.

| Задача | Пример для фамилии Иванов |
|---|---|
| 1 | `иванов_01.sh` |
| 2 | `иванов_02.sh` |
| 3 | `иванов_03.sh` |
| 4 | `иванов_04.sh` |


## Задача 1. Персональное приветствие — 2 баллa

**Файл:** `фамилия_01.sh`

Создайте скрипт, который:

1. Запрашивает имя пользователя.
2. Запрашивает возраст.
3. Выводит персональное приветствие.
4. Вычисляет, сколько лет пользователю исполнится через 5 лет.

**Пример работы:**

```text
Введите имя: Иван
Введите возраст: 18

Привет, Иван!
Сейчас тебе 18 лет.
Через 5 лет тебе будет 23 года.
```


## Задача 2. Генератор проекта — 2 балла

**Файл:** `фамилия_02.sh`

Создайте скрипт, который:

1. Запрашивает название проекта.
2. Создаёт папку с указанным названием.
3. Внутри создаёт три папки: `src`, `docs`, `tests`.
4. Создаёт файл `README.md`, содержащий название проекта.
5. Создаёт пустой файл `src/main.py`.
6. Выводит сообщение об успешном создании проекта.

**Пример структуры результата:**

```text
my_project/
├── README.md
├── src/
│   └── main.py
├── docs/
└── tests/
```


## Задача 3. Анализатор файла — 3 балла

**Файл:** `фамилия_03.sh`

Создайте скрипт, который:

1. Запрашивает путь к текстовому файлу.
2. Проверяет, существует ли файл.
3. Если файл существует, выводит количество **строк**, **слов** и **байтов**.
4. Если файла нет, выводит сообщение об ошибке.

Для решения можно использовать команду `wc`.

**Пример работы:**

```text
Введите путь к файлу: notes.txt

Файл найден!

Строк: 15
Слов: 120
Байтов: 856
```

*Числа приведены для примера и зависят от содержимого файла.*


## Задача 4. Резервное копирование — 4 балла

**Файл:** `фамилия_04.sh`

Создайте скрипт, который:

1. Запрашивает путь к папке.
2. Проверяет её существование.
3. Создаёт копию папки с суффиксом `_backup`.
4. Если папка резервной копии уже существует, сообщает об ошибке и не перезаписывает её.
5. Выводит сообщение об успешном копировании.

**Пример работы:**

```text
Введите путь к папке: my_project

Создаю резервную копию...

Готово!
Резервная копия: my_project_backup
```

**Важно:** исходная папка должна остаться без изменений.
