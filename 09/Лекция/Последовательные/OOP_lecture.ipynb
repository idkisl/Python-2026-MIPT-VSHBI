**Задание 4.**

Напишите человеку метод, который здоровается с конкретным другом.

```python
class Human:
    def hello(self):
        print("Hello!")

    def hello_person(self, name):
        print(f"Hello, {name}")


d = Human()
d.hello()
d.hello_person("Ivan")
```
