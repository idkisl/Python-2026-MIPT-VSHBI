## 1. Аннотации типов

Аннотации типов помогают читать сигнатуру функции и понимать, какие данные она ожидает и что возвращает.

Общий вид:

```python
def function_name(arg1: Type1, arg2: Type2) -> ReturnType:
    ...
```

Важно: Python по умолчанию **не проверяет** аннотации типов во время выполнения программы.

````python
def is_even(x: int) -> bool:
    return x % 2 == 0

print(is_even(10))
print(is_even(7))
````

````python
def average(values: list[float]) -> float:
    return sum(values) / len(values)

def min_max(values: list[int]) -> tuple[int, int]:
    return min(values), max(values)

print(average([1.5, 2.5, 4.0]))
print(min_max([7, 1, 9, 3]))
````

### Аннотация — это не запрет

Следующая программа выполняется, хотя аннотация говорит, что функция ожидает `int`.

````python
def double(x: int) -> int:
    return x * 2

print(double("abc"))
````

## 2. Функция как объект

В Python функция — это объект.

Поэтому функцию можно:

- сохранить в переменную;
- положить в список;
- сохранить в словарь;
- передать другой функции;
- вернуть из функции.

````python
def square(x):
    return x * x

print(square)
print(type(square))
print(square(5))
````

## 3. Присваивание функции переменной

Сравните:

```python
f = square
```

и

```python
f = square()
```

В первом случае мы сохраняем **саму функцию**.  
Во втором — сначала вызываем функцию и сохраняем **результат её работы**.

````python
def square(x):
    return x * x

f = square

print(f(6))
print(f is square)
````

## 4. Функции в списках и словарях

````python
def add(a, b):
    return a + b

def sub(a, b):
    return a - b

def mul(a, b):
    return a * b

operations = [add, sub, mul]

for operation in operations:
    print(operation(10, 3))
````

````python
operations = {
    "+": add,
    "-": sub,
    "*": mul,
}

print(operations["+"](5, 7))
print(operations["*"](5, 7))
````

Так можно строить таблицы операций без длинных цепочек `if/elif`.

## 5. `lambda` — короткая функция

`lambda` создаёт объект-функцию без отдельного `def`.

Общий вид:

```python
lambda аргументы: выражение
```

У `lambda` может быть только одно выражение.

````python
square = lambda x: x * x
maximum = lambda a, b: a if a > b else b

print(square(7))
print(maximum(10, 4))
````

````python
operations = {
    "+": lambda a, b: a + b,
    "-": lambda a, b: a - b,
    "*": lambda a, b: a * b,
}

print(operations["*"](8, 9))
````

`lambda` особенно полезна, когда функция:

- короткая;
- нужна в одном месте;
- используется как аргумент другой функции.

# Функции высшего порядка

## 6. Передача функции другой функции

**Функция высшего порядка (Higher-Order Function, HOF)** — функция, которая:

- принимает функцию как аргумент;
- и/или возвращает функцию.

````python
def apply(f, x):
    return f(x)

def cube(x):
    return x ** 3

print(apply(cube, 4))
print(apply(lambda x: x + 10, 4))
````

Главная идея:

> функцию можно передавать как данные.

## 7. Предикаты и преобразования

### Предикат

Предикат — функция, возвращающая `bool`.

Примеры:

```python
lambda x: x > 0
lambda x: x % 2 == 0
lambda s: len(s) >= 5
```

### Преобразование

Преобразование получает значение и возвращает его новую форму.

```python
lambda x: x ** 2
lambda x: 2 * x + 1
lambda s: s.upper()
```

````python
is_positive = lambda x: x > 0
square = lambda x: x ** 2

print(is_positive(-3))
print(is_positive(7))
print(square(5))
````

# 8. `map`, `filter`, `zip`

## `map`

`map(function, iterable)` применяет функцию к каждому элементу последовательности.

````python
numbers = [1, 2, 3, 4, 5]

result = map(lambda x: x ** 2, numbers)

print(result)
print(list(result))
````

`map` возвращает не список, а специальный объект, по которому можно итерироваться.

````python
words = ["python", "cat", "algorithm", "code"]

print(list(map(len, words)))
print(list(map(str.upper, words)))
````

## `filter`

`filter(predicate, iterable)` оставляет только те элементы, для которых предикат возвращает `True`.

````python
numbers = [-4, -1, 0, 2, 5, 8]

positive = filter(lambda x: x > 0, numbers)
even = filter(lambda x: x % 2 == 0, numbers)

print(list(positive))
print(list(even))
````

````python
words = ["python", "", "cat", "algorithm", "a"]

print(list(filter(lambda s: len(s) >= 4, words)))
````

## `zip`

`zip` объединяет элементы нескольких последовательностей по одинаковым позициям.

````python
names = ["Анна", "Борис", "Вера"]
scores = [80, 91, 75]

pairs = list(zip(names, scores))
print(pairs)
````

````python
a = [1, 2, 3]
b = [10, 20, 30]

sums = list(map(lambda pair: pair[0] + pair[1], zip(a, b)))
print(sums)
````

`zip` не является функцией высшего порядка, но часто используется рядом с `map` и `filter` при обработке данных.

# 8. `key` в `sorted`, `min`, `max`

Параметр `key` принимает функцию, которая говорит Python:

> какое значение использовать для сравнения элементов.

````python
words = ["python", "cat", "algorithm", "code"]

print(sorted(words, key=len))
print(max(words, key=len))
print(min(words, key=len))
````

````python
numbers = [-10, 2, -3, 7, 1]

print(sorted(numbers, key=abs))
print(max(numbers, key=abs))
````

````python
points = [(2, 5), (-1, 8), (3, 1), (0, 4)]

print(min(points, key=lambda p: p[1]))
print(sorted(points, key=lambda p: p[0] ** 2 + p[1] ** 2))
````

### Сложный ключ

Ключ может быть кортежем.

````python
students = [
    ("Иванов", 4),
    ("Петров", 5),
    ("Сидоров", 4),
    ("Алексеев", 5),
]

result = sorted(students, key=lambda x: (-x[1], x[0]))
print(result)
````

Сначала сравнивается первый элемент ключа, затем второй.

```python
(-оценка, фамилия)
```

даёт:

1. оценку по убыванию;
2. фамилию по возрастанию.

# 8. Собственные функции высшего порядка

## `transform`

Аналог идеи `map`.

````python
def transform(data, f):
    result = []

    for x in data:
        result.append(f(x))

    return result

print(transform([1, 2, 3, 4], lambda x: x ** 2))
````

## `select`

Аналог идеи `filter`.

````python
def select(data, predicate):
    result = []

    for x in data:
        if predicate(x):
            result.append(x)

    return result

print(select([-3, -1, 0, 2, 5], lambda x: x > 0))
````

## `count_if`

Подсчитать количество элементов, удовлетворяющих предикату.

````python
def count_if(data, predicate):
    count = 0

    for x in data:
        if predicate(x):
            count += 1

    return count

print(count_if([1, 2, 3, 4, 5, 6], lambda x: x % 2 == 0))
````

# 8. Функция, возвращающая функцию

Функция высшего порядка может не только принимать, но и **создавать и возвращать функции**.

````python
def make_multiplier(k):
    def multiply(x):
        return k * x

    return multiply

double = make_multiplier(2)
triple = make_multiplier(3)

print(double(10))
print(triple(10))
````

# 8. Простые замыкания

В примере выше внутренняя функция `multiply` продолжает помнить значение `k` даже после завершения `make_multiplier`.

Такое поведение называется **замыканием**.

````python
def make_power(power):
    def raise_to_power(x):
        return x ** power

    return raise_to_power

square = make_power(2)
cube = make_power(3)

print(square(5))
print(cube(5))
````

````python
def make_between(left, right):
    def predicate(x):
        return left <= x <= right

    return predicate

between_10_20 = make_between(10, 20)

print(between_10_20(15))
print(between_10_20(25))
````

# 8. Обёртка функции

Предположим, мы хотим выполнить некоторые действия до и после функции.

````python
def hello():
    print("Hello")

def wrapper():
    print("Start")
    hello()
    print("Finish")

wrapper()
````

Проблема: `wrapper` жёстко связан с `hello`.

Сделаем обёртку универсальной.

````python
def decorate(func):
    def wrapper():
        print("Start")
        func()
        print("Finish")

    return wrapper

def hello():
    print("Hello")

wrapped_hello = decorate(hello)
wrapped_hello()
````

# 8. `func = decorator(func)`

Можно заменить исходную функцию её обёрнутой версией.

````python
def decorator(func):
    def wrapper():
        print("Before")
        result = func()
        print("After")
        return result

    return wrapper

def greet():
    print("Hi!")

greet = decorator(greet)

greet()
````

---

# 8. Синтаксис `@decorator`

Python предоставляет короткую запись:

```python
@decorator
def f():
    ...
```

Это примерно эквивалентно:

```python
f = decorator(f)
```

````python
def decorator(func):
    def wrapper():
        print("Before")
        result = func()
        print("After")
        return result

    return wrapper

@decorator
def greet():
    print("Hello!")

greet()
````

# 8. `*args`, `**kwargs` в декораторах

Чтобы декоратор подходил для функций с любым количеством аргументов, используем уже знакомые `*args` и `**kwargs`.

````python
def logger(func):
    def wrapper(*args, **kwargs):
        print("args:", args)
        print("kwargs:", kwargs)

        result = func(*args, **kwargs)

        print("result:", result)
        return result

    return wrapper

@logger
def add(a, b):
    return a + b

add(5, 7)
````

# 8. Практические декораторы


````python
def simple_decorator(func):
    def wrapper(*args, **kwargs):
        return func(*args, **kwargs)

    return wrapper

@simple_decorator
def add(a, b):
    "Складывает два числа."
    return a + b

print(add.__name__)
print(add.__doc__)
````

````python

def good_decorator(func):
    def wrapper(*args, **kwargs):
        return func(*args, **kwargs)

    return wrapper

@good_decorator
def add(a, b):
    "Складывает два числа."
    return a + b

print(add.__name__)
print(add.__doc__)
````
