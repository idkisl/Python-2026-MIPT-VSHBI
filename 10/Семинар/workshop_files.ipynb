## Подготовка файлов

```python
import csv
import json

with open("text.txt", "w", encoding="utf-8") as f:
    f.write("Python — язык программирования.\n")
    f.write("Мы учимся работать с файлами.\n")
    f.write("Python удобен для обработки данных.\n")

with open("numbers.txt", "w", encoding="utf-8") as f:
    f.write("12\n7\n20\n5\n16\n")

with open("students.txt", "w", encoding="utf-8") as f:
    f.write("Анна 5\nБорис 4\nВера 5\nГлеб 3\nДаша 4\n")

with open("products.csv", "w", encoding="utf-8", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["product", "price"])
    writer.writerows([["Хлеб", 70], ["Молоко", 110], ["Яблоки", 160], ["Сыр", 250]])

with open("person.json", "w", encoding="utf-8") as f:
    json.dump({"name": "Анна", "age": 19, "city": "Москва"}, f, ensure_ascii=False, indent=4)

print("Файлы для семинара созданы!")
```
