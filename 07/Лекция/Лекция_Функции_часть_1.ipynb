# 1. Зачем нужны функции

Представим, что одну и ту же последовательность команд нужно выполнить несколько раз.

Можно каждый раз копировать код.

Но удобнее один раз дать этому действию имя — создать **функцию**.

````python
print("Привет!")
print("Привет!")
print("Привет!")
````

То же самое с функцией:

````python
def hello():
    print("Привет!")

hello()
hello()
hello()
hello()
````

Функция состоит из двух важных частей:

```python
def hello():
    print("Привет!")
```

Это **объявление функции**.

А запись

```python
hello()
```

— это **вызов функции**.

Пока функцию не вызвали, команды внутри неё не выполняются.

# 2. Параметры функции

Часто функция должна работать с разными данными.

Например, функция может приветствовать разных людей.

````python
def say_hello(name):
    print("Привет,", name)


say_hello("Анна")
say_hello("Иван")
say_hello("Пётр")
````

В записи

```python
def say_hello(name):
```

`name` — **параметр** функции.

В вызове

```python
say_hello("Анна")
```

`"Анна"` — **аргумент**.

Можно запомнить так:

- параметр — переменная в объявлении функции;
- аргумент — конкретное значение при вызове.

# 4. `print` и `return`

Это одна из самых важных тем.

`print()` просто **показывает значение на экране**.

`return` **возвращает значение из функции**.

````python
def square_print(x):
    print(x ** 2)


def square_return(x):
    return x ** 2

a = square_return(5)
square_print(5)
print(a)
````

Почему `return` важен?

Потому что возвращённое значение можно сохранить и использовать дальше.

````python
def square(x):
    return x ** 2


a = square(5)
b = square(10)

print(a)
print(b)
print(a + b)
````

С функцией, которая только печатает значение, так сделать не получится.

````python
def square_print(x):
    print(x ** 2)


a = square_print(5)

print("В переменной a:", a)
````

Если функция ничего явно не возвращает, Python возвращает `None`.

# 5. `return` завершает функцию

Как только Python выполняет `return`, функция сразу заканчивает работу.

````python
def sign(x):
    if x > 0:
        return 1

    if x < 0:
        return -1

    return 0


print(sign(10))
print(sign(-5))
print(sign(0))
````

Здесь после выполнения одного из `return` остальные команды функции уже не выполняются.

# 6. Функция внутри выражения

Функцию с `return` можно использовать почти так же, как обычное значение.

````python
def square(x):
    return x ** 2


print(square(4) + square(3))
````

````python
def is_even(x):
    return x % 2 == 0


if is_even(10):
    print("Число чётное")
````

Это очень важная идея:

```python
square(4)
```

после выполнения функции превращается в её результат.

# 7. Локальные переменные

Переменная, созданная внутри функции, обычно используется только внутри этой функции.

````python
def f(x):
    y = x + 10
    print(y)


f(5)
print(y, x)

````

Здесь `x` и `y` относятся к работе функции `f`.

Для начала достаточно простого правила:

> Всё, что нужно функции для работы, лучше передавать через параметры.  
> Результат работы лучше возвращать через `return`.

Напишите функцию

```python
sum_squares(a, b)
```

которая возвращает

```text

```

````python
def sum_squares(a, b):
    return a ** 2 + b ** 2
````

````python
def is_square_matrix(matrix):
    N = len(matrix)
    for i in range(len(matrix)):
        if len(matrix[i]) != N:
            return False
    return True
    

print(is_square_matrix([[1, 2], [3, 4]]))
print(is_square_matrix([[1, 2], [3, 4], [5, 6]]))
print(is_square_matrix([[1, 2, 3], [3, 4, 7]]))
print(is_square_matrix([[1, 2, 3], [3, 4]]))
````

````python
def determinant(matrix):
    if not is_square_matrix(matrix):
        return "Это не квадратная матрица"
    if len(matrix) != 2:
        return "Не умеем считать детерминант большой матрицы"
    return matrix[0][0] * matrix[1][1] - matrix[0][1] * matrix[1][0]


print(determinant([[1, 2], [3, 4]]))



````
