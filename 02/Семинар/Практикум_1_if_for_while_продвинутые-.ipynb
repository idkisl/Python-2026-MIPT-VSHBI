# Практикум Python: `if`, `for`, `while`


## Блок 1. Условный оператор `if`


**Задача 1. Температура тела.**  
Вводится температура тела. Выведите `пониженная`, `норма` или `повышенная` по тем же правилам, что и в первом практикуме.


**Задача 2. pH раствора.**  
Вводится значение pH. Определите тип среды: кислая, нейтральная или щелочная.


**Задача 3. Стоп-кодон.**  
Вводится кодон длины 3. Определите, является ли он одним из стоп-кодонов `TAA`, `TAG`, `TGA`.


**Задача 4. Комплементарность.**  
Вводятся два нуклеотида. Определите, являются ли они комплементарной парой ДНК.


**Задача 5. Рост колонии бактерий.**  
Вводится начальное количество бактерий `N`. В течение 10 часов каждый час количество удваивается. Выведите численность колонии после каждого часа. Используйте `for`.


**Задача 6. Измерения температуры.**  
Вводится число измерений `N`, затем `N` температур. Посчитайте, сколько значений выше 37.0 °C.


**Задача 7. Подсчёт нуклеотидов.**  
Вводится строка ДНК. Посчитайте количество символов `A`, `T`, `G`, `C`. Используйте цикл `for`.


**Задача 8. GC-состав последовательности.**  
Вводится строка ДНК. Найдите процент нуклеотидов `G` и `C` среди всех символов строки.


**Задача 9. Поиск стоп-кодонов.**  
Вводится строка, состоящая из нескольких кодонов без пробелов. Просмотрите строку по три символа и посчитайте количество кодонов `TAA`, `TAG`, `TGA`.


**Задача 10. Химический эксперимент.**  
Вводится `N`, затем `N` значений pH. Посчитайте отдельно количество кислых, нейтральных и щелочных растворов.


**Задача 11. Средняя скорость реакции.**  
Вводится `N`, затем `N` значений скорости реакции. Найдите среднее арифметическое и максимальное значение.


**Задача 12. Фильтрация результатов секвенирования.**  
Вводится `N`, затем `N` длин прочтений ДНК. Посчитайте, сколько прочтений имеют длину не меньше 100 и не больше 300.


**Задача 13. Частота мутаций.**  
Вводится `N`, затем `N` пар символов: исходный и новый нуклеотид. Посчитайте количество случаев, где символы различаются.


**Задача 14. Самое длинное название вида.**  
Вводится `N`, затем `N` названий организмов. Выведите самое длинное название.


**Задача 15. Деление бактерий до порога.**  
Вводятся начальное количество бактерий и предельное значение. Колония удваивается каждый час. Определите, через сколько часов она достигнет или превысит заданный предел. Используйте `while`.


**Задача 16. Стабилизация температуры.**  
Вводятся показания температуры до тех пор, пока не будет введено значение от 36.5 до 37.0 включительно. Посчитайте количество предыдущих измерений.


**Задача 17. Завершение эксперимента.**  
Вводятся значения pH до тех пор, пока не будет введено `7`. Посчитайте, сколько среди предыдущих значений было кислых.


**Задача 18. Сигнал секвенатора.**  
Вводятся строки ДНК до тех пор, пока не будет введена пустая строка. Посчитайте количество последовательностей, содержащих `ATG`.


**Задача 19. Контроль загрязнения.**  
Вводятся концентрации вещества до тех пор, пока не будет введено отрицательное число. Найдите максимальную концентрацию среди введённых до сигнала окончания.


**Задача 20. Поиск первого стоп-кодона.**  
Вводятся кодоны по одному. Ввод заканчивается, когда встретится `TAA`, `TAG` или `TGA`. Посчитайте, сколько обычных кодонов было введено до первого стоп-кодона.


**Задача 21. Накопление вещества.**  
В сосуде изначально находится `A` граммов вещества. Каждый шаг добавляется `B` граммов. Определите минимальное число шагов, после которого масса станет не меньше `M`.


**Задача 22. Последовательность измерений.**  
Вводятся целые числа до `0`. Найдите минимальное и максимальное значение среди введённых чисел.


## Блок 2. Смешанные задачи


**Задача 23. Анализ ДНК-фрагмента.**  
Вводится строка ДНК. Посчитайте число `G` и `C`. Если GC-состав больше 60% — выведите `высокий GC`, если от 40% до 60% включительно — `средний GC`, иначе — `низкий GC`. Используйте `for` и `if`.


**Задача 24. Контроль температуры реактора.**  
Вводится `N`, затем `N` температур. Для каждого значения выведите `низкая`, `норма` или `высокая`. В конце выведите число значений, вышедших за допустимый диапазон от 20 до 80 °C.


**Задача 25. Поиск мотива в ДНК.**  
Вводится строка ДНК и короткий мотив. Посчитайте, сколько раз мотив начинается в строке. Используйте цикл `for`.


**Задача 26. Модель эпидемии.**  
Вводится начальное число заболевших и число дней. Каждый день число заболевших увеличивается на 20%. Выведите прогноз на каждый день. Если число становится больше 1000, дополнительно выведите `высокий уровень`.


**Задача 27. Отбор проб.**  
Вводятся значения концентрации до тех пор, пока не будет введено `0`. Посчитайте, сколько проб имеют концентрацию выше 50. Если таких больше половины всех проб, выведите `повышенный уровень`, иначе `норма`.


**Задача 28. Кодирование аминокислот.**  
Вводится несколько кодонов по одному до строки `STOP`. Для кодонов `ATG` выводите `Met`, для `TTT` и `TTC` — `Phe`, для остальных — `другая аминокислота`. Посчитайте количество распознанных кодонов.


**Задача 29. Проверка результатов эксперимента.**  
Вводится `N`, затем `N` измерений. Измерение считается корректным, если оно находится в диапазоне от 0 до 100 включительно. Посчитайте количество корректных и некорректных результатов.


**Задача 30. Мини-биоинформатика: сравнение двух последовательностей.**  
Вводятся две строки ДНК одинаковой длины. Посчитайте число позиций, в которых символы различаются. Это расстояние Хэмминга.
