# Практикум Python: условный оператор `if`



## Блок 1. Базовые условия


**Задача 1. Трёхзначное число.**  
Дано натуральное число. Определите, является ли оно трёхзначным. Если является — выведите цифру сотен, иначе выведите `NO`.


**Задача 2. Температура тела.**  
Пользователь вводит температуру тела в градусах Цельсия.  
Если температура ниже 36.0 — выведите `пониженная`, от 36.0 до 37.0 включительно — `норма`, выше 37.0 — `повышенная`.


**Задача 3. Пульс после нагрузки.**  
Вводятся возраст человека и его пульс после физической нагрузки.  
Для упрощённой оценки считайте, что пульс:
- до 100 — низкая нагрузка;
- от 100 до 140 включительно — умеренная;
- выше 140 — высокая.  
Если возраст меньше 14 лет, дополнительно выведите `детский режим оценки`.


**Задача 4. pH раствора.**  
Вводится значение pH от 0 до 14.  
Если pH меньше 7 — выведите `кислая среда`, если равно 7 — `нейтральная`, если больше 7 — `щелочная`.


**Задача 5. Агрегатное состояние воды.**  
Вводится температура воды в градусах Цельсия при нормальном атмосферном давлении.  
Если температура меньше 0 — выведите `лёд`, от 0 до 100 — `жидкость`, выше 100 — `пар`.


**Задача 6. Скорость химической реакции.**  
Вводятся температура реакции и наличие катализатора (`да` или `нет`).  
Считайте, что реакция идёт быстро, если температура не ниже 60 °C или используется катализатор.  
Выведите `быстро` или `медленно`.


**Задача 7. Концентрация раствора.**  
Вводятся масса растворённого вещества и масса раствора.  
Если масса раствора равна 0 — выведите `ошибка`. Иначе вычислите массовую долю вещества в процентах и выведите:
- `разбавленный`, если меньше 10%;
- `средний`, если от 10% до 30% включительно;
- `концентрированный`, если больше 30%.


**Задача 8. Группа крови по системе ABO.**  
Пользователь вводит наличие антигенов A и B двумя строками (`да`/`нет`).  
Определите группу крови:
- нет A и нет B → `O`;
- есть A, нет B → `A`;
- нет A, есть B → `B`;
- есть A и B → `AB`.


**Задача 9. Определение типа мутации.**  
Вводятся два символа ДНК — исходный нуклеотид и новый (`A`, `T`, `G`, `C`).  
Если символы совпадают — выведите `мутации нет`. Иначе выведите `замена нуклеотида`.


**Задача 10. Проверка нуклеотида.**  
Вводится один символ. Если это одна из букв `A`, `T`, `G`, `C`, выведите `корректный нуклеотид`, иначе — `ошибка`.


**Задача 11. Комплементарная пара ДНК.**  
Вводятся два нуклеотида. Определите, являются ли они комплементарной парой.  
Пары `A–T` и `T–A`, а также `G–C` и `C–G` считаются комплементарными.


**Задача 12. GC-состав короткого фрагмента.**  
Вводятся три символа — три нуклеотида ДНК.  
Посчитайте без использования циклов, сколько среди них букв `G` или `C`.  
Если таких нуклеотидов 0 — выведите `низкий GC`, 1 или 2 — `средний GC`, 3 — `высокий GC`.


**Задача 13. Старт-кодон.**  
Вводится строка длины 3. Если она равна `ATG`, выведите `старт-кодон`, иначе — `не старт-кодон`.


**Задача 14. Стоп-кодон.**  
Вводится строка длины 3. Если она равна `TAA`, `TAG` или `TGA`, выведите `стоп-кодон`, иначе — `обычный кодон`.


**Задача 15. Содержание кислорода.**  
Вводится процент кислорода в газовой смеси.  
Если значение меньше 16 — выведите `опасно низкое`, от 16 до 23.5 включительно — `допустимое`, выше 23.5 — `повышенное`.


**Задача 16. Деление клетки.**  
Вводится число хромосом до деления и после деления.  
Если числа равны — выведите `число хромосом сохранилось`, если второе в два раза меньше первого — `возможное редукционное деление`, иначе — `другая ситуация`.


**Задача 17. Рост растения.**  
Вводятся высота растения неделю назад и сейчас.  
Если текущая высота больше — выведите `растение выросло`, если равна — `рост не изменился`, если меньше — `высота уменьшилась`.


**Задача 18. Классификация животного по температуре тела.**  
Вводится температура тела животного.  
Для учебной модели считайте:
- ниже 25 °C — `низкая`;
- от 25 до 38 °C включительно — `средняя`;
- выше 38 °C — `высокая`.  
Выведите соответствующую категорию.


**Задача 19. Электронный пропуск в лабораторию.**  
Вводятся текущий час и возраст посетителя.  
Лаборатория открыта с 8 до 20 часов включительно. Посетители младше 14 лет могут находиться в лаборатории только до 18 часов включительно.  
Выведите `доступ разрешён` или `доступ запрещён`.


**Задача 20. Високосный год.**  
Вводится номер года. Определите, является ли год високосным.  
Год високосный, если делится на 400 или делится на 4, но не делится на 100.
