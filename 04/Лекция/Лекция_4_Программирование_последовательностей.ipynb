# Программирование последовательностей
### Лекция 4 · Python · 1 курс

**Главный вопрос занятия:**  
> Что действительно нужно помнить о уже просмотренной части последовательности?

# Главная идея

Пусть элементы приходят по одному.

Необязательно хранить всю последовательность.

Часто достаточно нескольких переменных:

```python
total = 0
count = 0
best = ...
previous = ...
current_len = 1
best_len = 1
```

**Задача алгоритмиста — определить минимальное состояние.**

# 1. Накопитель

### Задача
Дано `n`, затем `n` целых чисел. Найдите их сумму.

Идея: хранить только накопленную сумму.

````python
n = int(input())
total = 0

for _ in range(n):
    x = int(input())
    total += x

print(total)
````

# 2. Счётчик

### Задача
Посчитать количество положительных элементов.

Счётчик изменяется только тогда, когда условие выполнено.

````python
n = int(input())
count = 0

for _ in range(n):
    x = int(input())
    if x > 0:
        count += 1

print(count)
````

# 3. Максимум последовательности

Ошибка:

```python
best = 0
```

Она ломает решение, если все числа отрицательные.

Надёжнее взять в качестве начального максимума **первый элемент**.

````python
n = int(input())
best = int(input())

for _ in range(n - 1):
    x = int(input())
    if x > best:
        best = x

print(best)
````

# 4. Максимум и его позиция

Теперь недостаточно помнить только значение.

Нужно хранить:

- лучший элемент;
- позицию, где он встретился.

````python
n = int(input())

best = int(input())
best_pos = 1

for pos in range(2, n + 1):
    x = int(input())

    if x > best:
        best = x
        best_pos = pos

print(best, best_pos)
````

# 5. Условный экстремум

### Задача
Найти минимальный положительный элемент.

Проблема: подходящего элемента может вообще не быть.

Удобный приём:

```python
best = None
```

````python
n = int(input())
best = None

for _ in range(n):
    x = int(input())

    if x > 0 and (best is None or x < best):
        best = x

print("NO" if best is None else best)
````

# Задача: количество возрастаний

Посчитать, сколько раз следующий элемент больше предыдущего.

Пример:

```text
3  5  2  7  8
   ↑     ↑  ↑
```

Ответ: `3`.

````python
n = int(input())
previous = int(input())
count = 0

for _ in range(n - 1):
    current = int(input())

    if current > previous:
        count += 1

    previous = current

print(count)
````

# 7. Максимальный скачок

### Задача

Найти

$
\max |a_i-a_{i-1}|
$

Нужно помнить только предыдущий элемент и лучший найденный ответ.

````python
n = int(input())
previous = int(input())
best = 0

for _ in range(n - 1):
    current = int(input())

    best = max(best, abs(current - previous))
    previous = current

print(best)
````

# 8. Серии подряд идущих элементов

Типовой шаблон:

```python
if серия_продолжается:
    current_len += 1
else:
    current_len = 1

best_len = max(best_len, current_len)
```

Нужны две величины:

- `current_len` — длина текущей серии;
- `best_len` — лучшая длина.

# Задача: одинаковые подряд

Найти максимальное количество одинаковых элементов, идущих подряд.

Пример:

```text
1 1 3 3 3 2 2 5
```

Ответ: `3`.

````python
n = int(input())
previous = int(input())

current_len = 1
best_len = 1

for _ in range(n - 1):
    current = int(input())

    if current == previous:
        current_len += 1
    else:
        current_len = 1

    best_len = max(best_len, current_len)
    previous = current

print(best_len)
````

# Задача: возрастающая серия

Найти длину самого длинного непрерывного строго возрастающего фрагмента.

```text
4  6  8  2  5  9  11  3
      3        4
```

Ответ: `4`.

````python
n = int(input())
previous = int(input())

current_len = 1
best_len = 1

for _ in range(n - 1):
    current = int(input())

    if current > previous:
        current_len += 1
    else:
        current_len = 1

    best_len = max(best_len, current_len)
    previous = current

print(best_len)
````

# 9. Локальный максимум

Элемент `a[i]` — локальный максимум, если

$
a_{i-1} < a_i > a_{i+1}
$

Здесь нужны уже **три соседних элемента**:

```python
left, middle, right
```

````python
n = int(input())

left = int(input())
middle = int(input())
count = 0

for _ in range(n - 2):
    right = int(input())

    if left < middle > right:
        count += 1

    left, middle = middle, right

print(count)
````

# 10. Скользящее окно

Иногда анализируем не отдельный элемент, а небольшой непрерывный фрагмент.

### Пример
Найти максимальную сумму трёх соседних элементов.

Окно:

```text
[a b c] d e f
 a [b c d] e f
 a b [c d e] f
```

````python
n = int(input())

a = int(input())
b = int(input())
c = int(input())

best = a + b + c

for _ in range(n - 3):
    a, b = b, c
    c = int(input())
    best = max(best, a + b + c)

print(best)
````

# 11. Первый повторившийся элемент

Здесь удобно множество `seen`.

Оно отвечает на вопрос:

> встречали ли мы этот элемент раньше?

````python
a = list(map(int, input().split()))

seen = set()
answer = None

for x in a:
    if x in seen:
        answer = x
        break

    seen.add(x)

print("NO" if answer is None else answer)
````

# 12. Частоты элементов

Если нужно знать количество появлений каждого значения, используем словарь:

```python
freq[x] = freq.get(x, 0) + 1
```

````python
a = list(map(int, input().split()))

freq = {}

for x in a:
    freq[x] = freq.get(x, 0) + 1

for value, count in freq.items():
    print(value, count)
````

# 13. Непрерывный фрагмент ≠ подпоследовательность

### Непрерывный фрагмент

```text
3  7  [2  5  8]  1
```

Все элементы стоят подряд.

### Подпоследовательность

```text
3  7  2  5  8  1
^     ^     ^
```

Элементы можно пропускать, но порядок сохраняется.

# 14. Максимальная сумма непрерывного фрагмента

Для каждого элемента спрашиваем:

> выгоднее начать новый фрагмент здесь  
> или продолжить предыдущий?

Переход:

$
current = \max(x,\ current+x)
$

````python
a = list(map(int, input().split()))

current = a[0]
best = a[0]

for x in a[1:]:
    current = max(x, current + x)
    best = max(best, current)

print(best)
````
