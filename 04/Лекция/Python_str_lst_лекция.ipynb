# Строки и списки в Python

### Индексы · срезы · перебор · методы · изменяемость

# 1. Строки `str`

Строка — это **последовательность символов**.

```python
word = "Python"
```

## Индексы

```text
Строка:  P  y  t  h  o  n
Индекс:  0  1  2  3  4  5
```

Индексация начинается с `0`.

````python
word = "Python"
print(word[0])
print(word[2])
print(word[-1])
````

## Отрицательные индексы

```text
 P   y   t   h   o   n
-6  -5  -4  -3  -2  -1
```

```python
word[-1]
```

— последний символ.

## Длина строки

```python
len(word)
```

Последний положительный индекс:

```python
len(word) - 1
```

````python
word = "Python"
print(len(word))
print(word[len(word) - 1])
````

## Перебор строки

По символам:

```python
for ch in word:
    print(ch)
```

По индексам:

```python
for i in range(len(word)):
    print(i, word[i])
```

## Когда нужны индексы?

Когда важно положение элемента:

- вывести индекс;
- сравнить соседние символы;
- обратиться к предыдущему или следующему символу.

## Срезы

```python
s[start:stop:step]
```

`start` включается, `stop` — нет.

```python
s[2:7]
s[4:]
s[:4]
s[:-1]
s[::2]
s[::-1]
```

````python
s = "abcdefghij"
print(s[2:7])
print(s[4:])
print(s[:4])
print(s[::-1])
````

## Проверка вхождения

```python
"Python" in text
"Java" not in text
```

`in` проверяет, встречается ли символ или подстрока.

````python
text = "Основы Python"
print("Python" in text)
print("Java" in text)
````

## Регистр

```python
s.lower()
s.upper()

ch.islower()
ch.isupper()
```

````python
s = "PyThOn"
print(s.lower())
print(s.upper())

for ch in s:
    if ch.isupper():
        print(ch)
````

## Сложение строк

```python
first = "Hello"
second = "Python"

result = first + " " + second
```

Создаётся **новая строка**.

## Сравнение строк

Можно использовать:

```python
==  !=  <  >  <=  >=
```

Сравнение идёт **посимвольно слева направо**.

```python
"apple" < "banana"
"cat" < "dog"
```

## Регистр влияет на сравнение

Если регистр не важен:

```python
a = input().lower()
b = input().lower()

print(a == b)
```

## Строки неизменяемы

Так нельзя:

```python
s = "cat"
s[0] = "b"
```

Вместо этого создаём новую строку:

```python
new_s = "b" + s[1:]
```

# 2. Списки `list`

Список — последовательность элементов.

```python
numbers = [10, 20, 30]
names = ["Анна", "Иван", "Олег"]
```

<div class="lesson-accent">
Список можно изменять.
</div>

## Создание списка

```python
a = []
a = [3, 7, 10]
```

Добавление в конец:

```python
a.append(15)
```

````python
a = []
a.append(10)
a.append(20)
a.append(30)
print(a)
````

## Индексы и изменение элемента

```python
a[0]
a[-1]
```

Список изменяем:

```python
a[1] = 100
```

````python
a = [10, 20, 30]
a[1] = 100
print(a)
````

## Длина и перебор

```python
len(a)
```

По значениям:

```python
for x in a:
    print(x)
```

По индексам:

```python
for i in range(len(a)):
    print(i, a[i])
```

## Срезы списка

```python
a[1:4]
a[::2]
a[::-1]
```

<div class="lesson-accent">
Срез создаёт новый список.
</div>

````python
a = [10, 20, 30, 40, 50, 60]
print(a[1:4])
print(a[::2])
print(a[::-1])
````

## Проверка вхождения

```python
15 in a
100 not in a
```

Работает так же, как для строк.

## Добавление: `append()`

```python
a.append(value)
```

`append()` изменяет существующий список.

## Удаление: `pop()`

```python
x = a.pop(1)
```

Удаляет по индексу и возвращает удалённый элемент.

```python
a.pop()
```

Удаляет последний элемент.

## Удаление: `remove()`

```python
a = [10, 20, 30, 20]
a.remove(20)
```

Удаляется **первое вхождение значения**.

## Удаление: `del`

```python
del a[0]
```

Коротко:

- `pop()` — по индексу, с возвратом;
- `remove()` — по значению;
- `del` — по индексу.

## Соседние элементы

```python
for i in range(1, len(a)):
    if a[i] > a[i - 1]:
        print(a[i])
```

Цикл начинается с `1`, потому что у `a[0]` нет предыдущего элемента.

# 3. Изменяемость коллекций

### Изменяем существующий список

```python
a = [1, 2, 3]
a.append(4)
```

Список `a` изменился.

## Создаём новую коллекцию

```python
a = [1, 2, 3]
b = a + [4]
```

Исходный `a` не меняется.

## Что меняет, а что создаёт?

### Меняют список
- `append()`
- `remove()`
- `pop()`
- `a[i] = ...`

### Создают новую коллекцию
- `a + b`
- `a[start:stop]`
- операции со строками

<div class="lesson-accent">
Строки — неизменяемы. Списки — изменяемы.
</div>

# 4. Практика

## Задача 5. Символы и индексы

Вводится строка. Для каждого символа выведите его индекс и сам символ.

## Задача 10. Заглавные буквы

Вводится строка. Выведите все символы, записанные в верхнем регистре.

## Задача 11. Сравнение слов

Вводятся два слова. Сравните их без учёта регистра.

Если слова совпадают, выведите `EQUAL`.

## Задача 12. Количество пробелов

## Задача 13. Склейка двух слов


## Задача 14. Создание списка

Вводится число `n`, затем `n` целых чисел.
Сохраните их в список и выведите получившийся список.

## Задача 15. Индексы списка

Вводится список из `n` чисел.

## Задача 16. Изменение элемента

Замените элемент с индексом `k` на `0` и выведите список.

## Задача 17. Добавление элемента

Вводится список из `n` чисел, затем ещё одно число `x`.

## Задача 19. Удаление по значению

Вводится список чисел и число `x`.
Гарантируется, что `x` встречается в списке.

Удалите первое вхождение `x` с помощью `remove()` и выведите список.

## Задача 20. Чётные индексы

## Задача 21. Сосед справа больше

## Задача 23. Проверка подписи

Вводится строка. Определите, есть ли в ней хотя бы одна заглавная и хотя бы одна строчная буква.

## Задача 24. Удаление запрещённого значения

Вводится `n` чисел и затем запрещённое число `x`.

Если его нет, список не изменяйте.

Выведите итоговый список.

## Задача 25. Список изменений температуры

Вводится `n` значений температуры по порядку наблюдений.

Создайте **новый список**, содержащий только те значения,
которые отличаются от непосредственно предыдущего значения.

Первое значение всегда должно попасть в новый список.

Например:

```text

```

Исходный список изменять не нужно.
