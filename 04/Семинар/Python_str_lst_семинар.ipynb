# Основы Python: строки и списки

## Семинар

# Часть 1. Строки

## Задача 1. Город

Дано название города. Определите, чётно или нечётно количество символов в нём.

Выведите `EVEN` или `ODD`.

## Задача 2. Эстафета слов

Даны два слова. Определите, заканчивается ли первое слово той же буквой,
с которой начинается второе.

Регистр не учитывайте.

## Задача 3. Зеркальная запись

Дано слово. Определите, одинаково ли оно читается слева направо и справа налево.

Регистр не учитывайте.

## Задача 4. Пропуск каждого второго символа

Дана строка. Сформируйте новую строку из символов с нечётными индексами.

## Задача 5. Первая половина

Дана строка чётной длины.
Поменяйте местами её левую и правую половины, создав новую строку.

Пример:

```text
abcdef -> defabc
```

## Задача 6. Контроль пароля

Вводится пароль.

Пароль считается подходящим, если:

- его длина не меньше 8;
- в нём есть хотя бы одна заглавная буква;
- в нём есть хотя бы одна строчная буква.

Выведите `YES` или `NO`.

## Задача 7. Раньше в алфавите

Вводятся три слова.
Выведите то, которое лексикографически меньше остальных.

Сравнение выполняйте без учёта регистра.

## Задача 8. Символ-разделитель

Вводятся три слова.
Сформируйте строку вида:

```text
слово1/слово2/слово3
```

Используйте сложение строк.

## Задача 9. Длинное слово

Вводится одно слово.
Если его длина не меньше 6 символов, выведите `LONG`, иначе — `SHORT`.

## Задача 10. Соседние одинаковые буквы

Дана строка.
Определите, есть ли в ней два одинаковых соседних символа.

Регистр учитывать.

# Часть 2. Списки

## Задача 11. Положительный сосед

Дан список целых чисел.
Выведите первый элемент, у которого и левый, и правый сосед положительны.

Первый и последний элементы не рассматриваются.

Если такого элемента нет, выведите `NO`.

## Задача 12. Локальный максимум

Дан список чисел.
Выведите все элементы, которые больше обоих своих соседей.

Первый и последний элементы не рассматриваются.

## Задача 13. Первая смена направления

Дан список чисел.
Найдите первый индекс `i`, для которого выполняется одно из условий:

- `a[i - 1] < a[i] > a[i + 1]`;
- `a[i - 1] > a[i] < a[i + 1]`.

Если такого элемента нет, выведите `NO`.

## Задача 14. Очистка журнала

В журнале записаны коды событий.
Дан список кодов и один код `x`.

Удалите **все** вхождения `x`, создавая новый список.
Исходный список не изменяйте.

## Задача 15. Сжатие последовательности

Дан список чисел.
Создайте новый список, в котором из каждой серии одинаковых подряд идущих значений
остаётся только одно.

Пример:

```text
[3, 3, 3, 5, 5, 2, 7, 7]
```

Результат:

```text
[3, 5, 2, 7]
```

## Задача 16. Пары соседей

Дан список чисел.
Выведите все соседние пары, сумма которых положительна.

Каждую пару выводите на отдельной строке.

## Задача 17. Лишний ноль

Дан список чисел.
Если в нём есть ноль, удалите **последний** ноль.

Если нулей нет, список не изменяйте.

## Задача 18. Новый элемент после больших

Дан список чисел и число `x`.

Создайте новый список из всех элементов исходного списка,
которые строго больше `x`.

Исходный список менять нельзя.

## Задача 19. Разворот участка

Дан список и два индекса `l` и `r`.

Создайте новый список, в котором элементы от `l` до `r` включительно
идут в обратном порядке, а остальные элементы остаются на своих местах.

Пример:

```text
[1, 2, 3, 4, 5, 6], l = 1, r = 4
```

Результат:

```text
[1, 5, 4, 3, 2, 6]
```

## Задача 20. Изменения курса

Дан список значений некоторой величины по времени.

Создайте новый список строк:

- `"UP"`, если текущее значение больше предыдущего;
- `"DOWN"`, если меньше;
- `"SAME"`, если равно.

Для списка из `n` значений в новом списке должно быть `n - 1` элементов.

# Часть 3. Строки + списки

## Задача 21. Самое длинное из трёх

Вводятся три слова.
Выведите самое длинное.

Если таких слов несколько, выведите первое из них.

## Задача 22. Фильтр списка фамилий

Вводится число `n`, затем `n` фамилий.
После этого вводится строка-фрагмент.

Создайте новый список только из тех фамилий,
которые содержат этот фрагмент без учёта регистра.

## Задача 23. Маршрут из трёх пунктов

Вводятся три названия пунктов маршрута.
Сформируйте строку вида:

```text
Москва -> Тверь -> Новгород
```

Используйте сложение строк.

## Задача 24. Повтор соседних символов

Вводится строка.
Определите, есть ли в ней два одинаковых символа, стоящих рядом.

Регистр учитывайте.

## Задача 25. Список участников

Вводится `n` строк с именами участников.
После этого вводится контрольное слово.

Создайте новый список имён, которые:

- содержат контрольное слово без учёта регистра;
- имеют длину не меньше длины контрольного слова + 3.

Выведите подходящие имена в исходном порядке.
