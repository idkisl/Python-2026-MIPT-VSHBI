# Функция как аргумент

## Задача 6. Функция `apply`

Напишите функцию:

```python
def apply(func, x):
    ...
```

Она должна применить переданную функцию `func` к значению `x`.

Проверьте её с функциями:

````python
def square(x):
    return x ** 2


def double(x):
    return x * 2
````

Ожидаемые примеры:

```python
apply(square, 5)   # 25
apply(double, 5)   # 10
```

## Задача 7. Вычислить два раза

Напишите функцию

```python
apply_twice(func, x)
```

которая дважды применяет функцию `func`.

Например, если:

```python
def plus_one(x):
    return x + 1
```

то:

```python
apply_twice(plus_one, 10)
```

должно вернуть `12`.

## Задача 8. Выбор операции

````python
def add_one(x):
    return x + 1


def double(x):
    return x * 2
````

Если:

```python
command = "add"
```

то в `operation` должна попасть функция `add_one`.

Если:

```python

```

то в `operation` должна попасть функция `double`.

После этого выполните:

