# 2. Именованные аргументы

Обычно аргументы передаются по порядку:

````python
def person(name, age):
    print(name, age)


person("Анна", 20)
````

Но можно явно написать, какому параметру передаётся значение:

````python
person(name="Анна", age=20)
person(age=20, name="Анна")
````

Такой способ называется передачей **по имени**.

Он особенно удобен, когда у функции много параметров.

# 4. Функция — это объект

Теперь новая идея.

Посмотрим на две записи:

````python
def square(x):
    return x ** 2


print(square)
print(square(5))
````

` square(5) ` — мы **вызываем** функцию.

` square ` без скобок — это сама функция.

То есть имя `square` хранит объект-функцию.

Это похоже на переменную:

````python
x = 10
print(x)
````

Только в переменной `x` лежит число, а в имени `square` — функция.

# 5. Функцию можно записать в другую переменную

````python
def square(x):
    return x ** 2


f = square

print(f(5))
print(f(10))
````

Мы не вызвали `square`.

Мы записали саму функцию в переменную `f`.

Поэтому теперь `f` тоже можно вызвать.

Очень важно не путать:

```python
f = square
```

— сохранить функцию.

```python
f = square(5)
```

— вызвать функцию и сохранить её результат.

# 6. Функцию можно передать другой функции

Обычные значения мы уже передавали:

````python
def square(x):
    return x ** 2
````

Но передать можно и саму функцию.

````python
def square(x):
    return x ** 2


def apply(func, x):
    return func(x)


print(apply(square, 5))
````

Разберём по шагам.

Вызов:

```python
apply(square, 5)
```

означает примерно:

```text
func = square
x = 5
```

А внутри выполняется:

```python
func(x)
```

то есть:

```python
square(5)
```

Можно передать другую функцию:

````python
def square(x):
    return x ** 2


def double(x):
    return x * 2


def apply(func, x):
    return func(x)


print(apply(square, 5))
print(apply(double, 5))
````

Получилась одна функция `apply`, которая умеет выполнять разные действия.

# 7. `lambda`

Иногда функция очень маленькая и нужна только в одном месте.

Например:

````python
def square(x):
    return x ** 2
````

Такую простую функцию можно записать короче:

````python
square = lambda x: x ** 2

print(square(5))
````

Запись

```python
lambda x: x ** 2
```

означает:

> функция получает `x` и возвращает `x ** 2`.

Сравнение:

````python
def square1(x):
    return x ** 2


square2 = lambda x: x ** 2


print(square1(5))
print(square2(5))
````

Для начала достаточно знать простую форму:

```python
lambda параметры: результат
```

Например:

````python
add = lambda a, b: a + b
is_even = lambda x: x % 2 == 0

print(add(3, 7))
print(is_even(10))
````

`lambda` не заменяет обычные функции.

Если функция длинная или сложная, лучше использовать `def`.

`lambda` удобно применять для совсем коротких действий.

# 8. Зачем `lambda` вообще нужна?

Чаще всего маленькую функцию не сохраняют отдельно, а сразу передают другой функции.

````python
def apply(func, x):
    return func(x)


print(apply(lambda x: x ** 2, 5))
print(apply(lambda x: x + 10, 5))
````

Читаем:

```python
apply(lambda x: x + 10, 5)
```

как:

> возьми число `5` и примени к нему функцию «прибавить 10».

# 9. Пример с `sorted`

`sorted()` умеет сортировать элементы.

Но иногда нужно объяснить, **по чему** сортировать.

````python
words = ["кот", "слон", "я", "программирование"]

print(sorted(words))
````

Например, отсортируем строки по длине:

````python
words = ["кот", "слон", "я", "программирование"]

print(sorted(words, key=len))
````

В `key` мы передали функцию `len`.

Важно: без скобок.

```python
key=len
```

а не

```python
key=len()
```

Можно использовать свою функцию:

````python
def last_digit(x):
    return x % 10


numbers = [32, 15, 41, 27, 10]

print(sorted(numbers, key=last_digit))
````

Или короткую `lambda`:

````python
numbers = [32, 15, 41, 27, 10]

print(sorted(numbers, key=lambda x: x % 10))
````

Это одно из самых полезных применений идеи «функция как объект».
