# Циклы в Python



### Лекция 1. Нумерация участников

В первой строке дано натуральное число `N`.

Выведите номера участников от `1` до `N`: каждый номер на отдельной строке.

**Ввод**
```text
5
```

**Вывод**
```text
1
2
3
4
5
```


### Лекция 2. Числа с заданным шагом

Даны три целых числа `A`, `B`, `K`. Известно, что `A ≤ B`, `K > 0`.

Выведите все числа от `A` до `B` включительно с шагом `K`.

**Ввод**
```text
3
15
4
```

**Вывод**
```text
3 7 11 15
```


### Лекция 3. Контроль температуры

В первой строке дано `N`, затем вводятся `N` целых значений температуры.

Нормальной считается температура от `18` до `25` включительно. Определите количество показаний вне нормы.

**Ввод**
```text
6
20
17
26
24
18
30
```

**Вывод**
```text
3
```


### Лекция 4. Лучший результат

Вводится число участников `N`, затем `N` целых результатов.

Выведите максимальный результат и номер участника, который получил его первым. Нумерация начинается с 1.

**Ввод**
```text
5
12
18
18
15
9
```

**Вывод**
```text
18 2
```


### Лекция 5. Наименьший делитель

Дано натуральное число `N > 1`.

Найдите его наименьший натуральный делитель, отличный от 1.

**Ввод**
```text
91
```

**Вывод**
```text
7
```


### Лекция 6. Обратный отсчёт

Дано натуральное число `N`.

Выведите числа от `N` до `1`, а затем строку `START`.

**Ввод**
```text
4
```

**Вывод**
```text
4
3
2
1
START
```


### Лекция 7. Количество цифр

Дано целое неотрицательное число `N`.

Определите количество цифр в его десятичной записи.

**Ввод**
```text
50720
```

**Вывод**
```text
5
```


### Лекция 8. Заряд робота

Робот начинает работу с зарядом `B` процентов. После каждого действия заряд уменьшается на 10% от текущего значения.

Определите, сколько действий робот сможет выполнить, пока перед очередным действием заряд не меньше `10`.

**Ввод**
```text
50
```

**Вывод**
```text
16
```


### Лекция 9. Поток измерений

Вводится последовательность целых чисел, заканчивающаяся нулём. Ноль не является элементом последовательности.

Найдите сумму положительных элементов и их количество.

**Ввод**
```text
5
-2
7
3
-1
0
```

**Вывод**
```text
15 3
```


### Лекция 10. Угадай число

Загаданное число равно `73`.

Пользователь вводит числа до тех пор, пока не угадает. После каждого неправильного ответа выведите `Больше`, если загаданное число больше введённого, и `Меньше` — иначе. После правильного ответа выведите количество попыток.

**Ввод**
```text
50
80
73
```

**Вывод**
```text
Больше
Меньше
3
```
