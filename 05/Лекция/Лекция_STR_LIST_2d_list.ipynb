# 1. Методы строк `str`

Методы строк позволяют выполнять типовые операции без ручного перебора символов.

## 1.1. Регистр

Основные методы:

- `upper()` — верхний регистр;
- `lower()` — нижний регистр;
- `capitalize()` — первый символ строки в верхний регистр;
- `title()` — каждое слово начинается с заглавной буквы.

````python
text = "python programming"

print(text.upper())
print(text.lower())
print(text.capitalize())
print(text.title())
````

Важно: методы строк возвращают новую строку.

````python
text = "python"

new_text = text.upper()

print(text)
print(new_text)
````

## 1.2. Поиск и подсчёт

Полезные методы:

- `count(sub)` — количество вхождений;
- `find(sub)` — индекс первого вхождения или `-1`;
- `startswith(prefix)` — проверка начала строки;
- `endswith(suffix)` — проверка конца строки.

````python
text = "abracadabra"

print(text.count("a"))
print(text.find("cad"))
print(text.startswith("abra"))
print(text.endswith("bra"))
````

## 1.3. Замена подстроки — `replace()`

```python
text.replace(old, new)
```

возвращает строку, в которой фрагмент `old` заменён на `new`.

````python
text = "10,20,30,40"

new_text = text.replace(",", "; ")

print(new_text)
````

# 2. `split()` и `join()`

Это одна из самых полезных связок при работе со строками и списками.

- `split()` превращает строку в список;
- `join()` собирает список строк обратно в строку.

## 2.1. `split()`

Без аргумента строка разбивается по пробельным символам.

````python
line = "Python  is   very useful"

words = line.split()

print(words)
````

Можно задать собственный разделитель.

````python
line = "red,green,blue,white"

colors = line.split(",")

print(colors)
````

Частый шаблон для ввода чисел:

````python
line = "10 20 -5 7"

parts = line.split()
numbers = []

print(parts)

for x in parts:
    numbers.append(int(x))

print(numbers)
````

## 2.2. `join()`

Общий вид:

```python
separator.join(strings)
```

````python
words = ["Python", "это", "удобно"]

print(" ".join(words))
print(", ".join(words))
print(" -> ".join(words))
````

Если исходные элементы — числа, сначала их нужно преобразовать в строки.

````python
numbers = [10, 20, 30]

for i in range(len(numbers)):
    numbers[i] = str(numbers[i])

result = " | ".join(numbers)

print(result)
````

# 3. Методы списка `list`

Переходим сразу к тем операциям со списком, которые часто используются в задачах.

## 3.1. Добавление элементов

### `append(x)`

Добавляет один элемент в конец.

````python
lst = [1, 2, 3]

lst.append(4)

print(lst)
````

### `extend(iterable)`

Добавляет сразу несколько элементов.

````python
lst = [1, 2, 3]

lst.extend([4, 5, 6])

print(lst)
````

### `insert(i, x)`

Вставляет элемент на позицию `i`.

````python
lst = [10, 30, 40]

lst.insert(1, 20)

print(lst)
````

## 3.2. Удаление элементов

- `remove(x)` — удалить первое вхождение значения `x`;
- `pop()` — удалить и вернуть последний элемент;
- `pop(i)` — удалить и вернуть элемент с индексом `i`;
- `clear()` — удалить все элементы.

````python
lst = [10, 20, 30, 20, 40]

lst.remove(20)
print(lst)

value = lst.pop(1)
print("Удалили:", value)
print(lst)
````

## 3.3. Поиск и подсчёт

- `count(x)` — число вхождений;
- `index(x)` — индекс первого вхождения.

````python
lst = [5, 7, 5, 9, 5, 1, 9]

print(lst.count(5))
print(lst.index(9))
````

## 3.4. Сортировка и разворот

- `sort()` — сортирует сам список;
- `reverse()` — разворачивает сам список;
- `sorted(lst)` — возвращает новый отсортированный список.

# 4. Списочные выражения

Списочное выражение позволяет построить новый список компактнее, чем обычный цикл.

```python
[expression for x in collection]
```

````python
numbers = [1, 2, 3, 4, 5]

squares = [x ** 2 for x in numbers]

lst = []
for x in numbers:
    lst.append(x ** 2)

print(squares)
````

Можно добавить условие.

````python
numbers = list(range(1, 11))

even = [x for x in numbers if x % 2 == 0]

print(even)
````

Можно одновременно преобразовать выбранные элементы.

````python
numbers = list(range(1, 11))

even_squares = [x ** 2 for x in numbers if x % 2 == 0]

print(even_squares)
````

# 5. Изменение списка внутри методов

При работе со списками важно помнить: многие методы изменяют сам список.

Например:

````python
lst = [3, 1, 2]

print("До:", lst)

result = lst.sort()

print("После:", lst)
print("Что вернул sort():", result)
````

Поэтому запись

```python
lst = lst.sort()
```

ошибочна: после неё `lst` станет `None`.

# 6. Двумерные списки

Двумерный список — это список списков.

Будем использовать его для хранения таблиц и матриц.

````python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print(matrix[0])
print(matrix[1][2])
````

Здесь:

```python
matrix[i]
```

— строка,

а

```python
matrix[i][j]
```

— отдельный элемент.

# 7. Создание матрицы

Например, матрица `3 × 4`, заполненная нулями:

````python
n = 3
m = 4

matrix = [[0 for j in range(m)] for i in range(n)]

for row in matrix:
    print(row)
````

Можно записать короче:

````python
n = 3
m = 4

matrix = [[0] * m for _ in range(n)]

print(matrix)
````

## Осторожно: неправильное создание матрицы

Такая запись опасна:

```python
matrix = [[0] * m] * n
```

В ней строки фактически ссылаются на один и тот же список.

````python
bad = [[0] * 3] * 3

bad[0][0] = 9

for row in bad:
    print(row)
````

# 9. Обход всех элементов

Для обработки всей матрицы используются два вложенных цикла.

````python
matrix = [
    [2, 7, 1],
    [8, 3, 5]
]

for i in range(len(matrix)):
    for j in range(len(matrix[i])):
        print(i, j, matrix[i][j])
````

# 12. Работа со столбцами

Столбец собирается из элементов с одинаковым вторым индексом.

````python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

column = [matrix[i][1] for i in range(len(matrix))]

print(column)
````

## Пример: максимумы по столбцам

````python
matrix = [
    [3, 8, 1],
    [7, 2, 9],
    [4, 6, 5]
]

n = len(matrix)
m = len(matrix[0])

result = []

for j in range(m):
    column = [matrix[i][j] for i in range(n)]
    result.append(max(column))

print(result)
````

# 13. Главная диагональ

Для квадратной матрицы:

```python
matrix[i][i]
```

— элементы главной диагонали.

````python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

main_diagonal = [matrix[i][i] for i in range(len(matrix))]

print(main_diagonal)
````

# 14. Побочная диагональ

Для квадратной матрицы размера `n × n`:

```python
matrix[i][n - 1 - i]
```

````python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

n = len(matrix)

secondary_diagonal = [
    matrix[i][n - 1 - i]
    for i in range(n)
]

print(secondary_diagonal)
````

# 15. Заполнение матрицы по координатам

Очень важный тип задач: значение элемента зависит от его позиции `i, j`.

## Пример: единичная матрица

````python
n = 5

matrix = []

for i in range(n):
    row = []
    for j in range(n):
        if i == j:
            row.append(1)
        else:
            row.append(0)
    matrix.append(row)

for row in matrix:
    print(*row)
````

## Пример: побочная диагональ и области относительно неё

Для побочной диагонали:

```python
i + j == n - 1
```

Выше:

```python
i + j < n - 1
```

Ниже:

```python
i + j > n - 1
```

````python
n = 5

matrix = []

for i in range(n):
    row = []

    for j in range(n):
        if i + j < n - 1:
            row.append(0)
        elif i + j == n - 1:
            row.append(1)
        else:
            row.append(2)

    matrix.append(row)

for row in matrix:
    print(*row)
````

# 16. Изменение элементов матрицы

## Пример: заменить отрицательные элементы на нули

````python
matrix = [
    [4, -2, 8],
    [-1, 5, -9],
    [7, 0, -3]
]

for i in range(len(matrix)):
    for j in range(len(matrix[i])):
        if matrix[i][j] < 0:
            matrix[i][j] = 0

for row in matrix:
    print(*row)
````
