# 1. Зачем нужны функции

Функция позволяет:

- вынести повторяющийся фрагмент программы;
- дать действию понятное имя;
- разбить большую задачу на небольшие части;
- повторно использовать уже написанный код.

Общий вид:

```python

```

````python
def hello():
    print("Привет!")


hello()
hello()
````

Функция сначала **определяется**, а затем **вызывается**.

Само объявление функции не выполняет её тело.

# 5. `print` и `return` — не одно и то же

`print()` показывает значение на экране.

`return` **возвращает результат из функции**, чтобы его можно было использовать дальше.

````python
def square_print(x):
    print(x ** 2)


def square_return(x):
    return x ** 2


a = square_print(5)
b = square_return(5)

print("a =", a)
print("b =", b)
````

У функции, которая ничего явно не возвращает, результатом является `None`.

# 7. `return` завершает функцию

После выполнения `return` функция сразу заканчивает работу.

````python
def sign(x):
    if x > 0:
        return 1

    if x < 0:
        return -1

    return 0


print(sign(10))
print(sign(-7))
print(sign(0))
````

# 9. Что происходит с переданными объектами

Если функция изменяет **сам список**, изменения видны и после завершения функции.

````python
def add_zero(lst):
    lst.append(0)
    
    
numbers = [1, 2, 3]

add_zero(numbers)

print(numbers)
````

Но если внутри функции просто присвоить параметру новый объект, внешняя переменная не изменится.

````python
def replace_list(lst):
    lst = [100, 200, 300]
    print("Внутри:", lst)


numbers = [1, 2, 3]

replace_list(numbers)

print("Снаружи:", numbers)
````

Для чисел и строк похожий эффект особенно заметен: операция создаёт новое значение, а внешняя переменная остаётся прежней.

````python
def increase(x):
    x += 10
    print("Внутри:", x)


a = 5

increase(a)

print("Снаружи:", a)
````

# 12. Именованные аргументы

До сих пор аргументы передавались в функцию в том же порядке, в котором записаны параметры.

Например:

````python
def person(name, age, city):
    print(name, age, city)


person("Анна", 20, "Москва")
````

Здесь используются **позиционные аргументы**:

- первое значение попадает в `name`;
- второе — в `age`;
- третье — в `city`.

Но аргумент можно передать явно по имени параметра.

````python
def person(name, age, city):
    print(name, age, city)


person(name="Анна", age=20, city="Москва")
````

Такой аргумент называется **именованным**.

Главное отличие: при именованной передаче мы явно указываем, какому параметру соответствует значение.

## Порядок именованных аргументов

Именованные аргументы можно передавать в другом порядке:

````python
def person(name, age, city):
    print(name, age, city)


person(city="Москва", name="Анна", age=20)
````

Порядок здесь не важен, потому что связь задаётся явно:

```python
name="Анна"
age=20
city="Москва"
```

# 13. Позиционные и именованные аргументы вместе

Можно сочетать два способа передачи аргументов:

````python
def person(name, age, city):
    print(name, age, city)


person("Анна", city="Москва", age=20)
````

Здесь:

- `"Анна"` — позиционный аргумент;
- `age=20` и `city="Москва"` — именованные.

Важно: **позиционные аргументы должны идти раньше именованных**.

Так делать можно:

```python
person("Анна", age=20, city="Москва")
```

А так нельзя:

```python
person(name="Анна", 20, "Москва")
```

# 15. Произвольное количество позиционных аргументов: `*args`

Иногда неизвестно заранее, сколько позиционных аргументов будет передано.

Тогда используется:

````python
def show_args(*args):
    print(args)


show_args()
show_args(10)
show_args(10, 20, 30)
show_args("a", "b", "c", "d")
````

Все дополнительные позиционные аргументы собираются в `args`.

Внутри функции `args` — кортеж.

По нему можно пройти циклом:

````python
def print_all(*args):
    for value in args:
        print(value)


print_all(5, 10, 15)
````

Можно использовать индексы:

````python
def first_and_last(*args):
    if len(args) == 0:
        return None

    return args[0], args[-1]


print(first_and_last(10, 20, 30, 40))
````

## Обычные параметры и `*args`

Можно сочетать обычные параметры и произвольное число дополнительных аргументов.

````python
def show_with_prefix(prefix, *values):
    for value in values:
        print(prefix, value)


show_with_prefix("Число:", 10, 20, 30)
````

В вызове:

```python
show_with_prefix("Число:", 10, 20, 30)
```

- `"Число:"` попадает в `prefix`;
- остальные значения собираются в `values`.

которая возвращает произведение всех переданных чисел.

# 16. Произвольное количество именованных аргументов: `**kwargs`

Если заранее неизвестно, какие именно именованные аргументы будут переданы, используется:

````python
def show_info(**kwargs):
    print(kwargs)


show_info(name="Анна", age=20)
show_info(city="Москва", course=1, group="A")
````

Все именованные аргументы собираются в словарь `kwargs`.

Например:

```python
{
    "name": "Анна",
    "age": 20
}
```

По `kwargs` можно пройти циклом:

````python
def show_info(**kwargs):
    for key in kwargs:
        print(key, "=", kwargs[key])


show_info(name="Анна", age=20, city="Москва")
````

Можно проверить наличие конкретного ключа:

````python
def describe(**kwargs):
    if "name" in kwargs:
        print("Имя:", kwargs["name"])

    if "age" in kwargs:
        print("Возраст:", kwargs["age"])


describe(name="Иван", age=19)
````

## Обычные параметры и `**kwargs`

Часть параметров может быть обязательной, а остальные именованные аргументы собираются в словарь.

````python
def create_user(login, **kwargs):
    print("Логин:", login)

    for key in kwargs:
        print(key, "=", kwargs[key])


create_user("ivan123", age=20, city="Москва")
````

# 17. `*args` и `**kwargs` вместе

# 18. Распаковка аргументов через `*`

Если позиционные аргументы уже лежат в списке или кортеже, их можно распаковать при вызове функции.

````python
def add_three(a, b, c):
    return a + b + c


numbers = [10, 20, 30]

print(add_three(*numbers))
print(*numbers)
````

Запись:

```python
add_three(*numbers)
```

эквивалентна:

```python
add_three(10, 20, 30)
```

Это удобно и для функций с `*args`:

````python
def total(*numbers):
    result = 0

    for x in numbers:
        result += x

    return result


values = [5, 7, 12, 3]

print(total(*values))
````

# 19. Распаковка словаря через `**`

Если именованные аргументы находятся в словаре, его можно распаковать:

````python
def person(name, age, city):
    print(name, age, city)


data = {
    "name": "Анна",
    "age": 20,
    "city": "Москва"
}

person(**data)
````

Здесь ключи словаря должны совпадать с именами параметров функции.

## Одновременная распаковка

Можно передать позиционные аргументы через `*`, а именованные — через `**`.

````python
def show(a, b, c, city, age):
    print(a, b, c, city, age)


numbers = [1, 2, 3]

info = {
    "city": "Москва",
    "age": 20
}

show(*numbers, **info)
````

## Что важно не путать

### В объявлении функции

```python
def f(*args, **kwargs):
```

звёздочки **собирают** аргументы.

### В вызове функции

```python
f(*values, **data)
```

звёздочки **распаковывают** последовательность и словарь.

# 20. Рекурсия

Только теперь переходим к рекурсии.

Рекурсия — это ситуация, когда функция вызывает саму себя.

Нужно следить за тремя вещами:

1. как изменяется аргумент;
2. где находится граничное условие;
3. приближается ли аргумент к границе.

## Нисходящая рекурсия

Если аргумент уменьшается:

```text
5 → 4 → 3 → 2 → 1 → 0
```

граничное условие располагается снизу.

````python
def countdown(n):
    if n == 0:
        print(0)
        return

    print(n)
    countdown(n - 1)


countdown(5)
````

Здесь аргумент изменяется так:

```python
n - 1
```

а граница:

```python
n == 0
```

## Восходящая рекурсия

Если аргумент увеличивается:

```text
0 → 1 → 2 → 3 → 4 → 5
```

граничное условие располагается сверху.

````python
def count_up(n, limit):
    print(n)

    if n == limit:
        return

    count_up(n + 1, limit)


count_up(0, 5)
````

Главное:

- аргумент уменьшается → нижняя граница;
- аргумент увеличивается → верхняя граница.

# 21. Рекурсивное вычисление

Оставим один основной вычислительный пример — факториал.

````python
def factorial(n):
    print(".", end="")
    if n == 0:
        return 1
    
    return n * factorial(n - 1)

a = factorial(400)
print()
print(a)
````

````python
def f(n):
    print(".", end="")
    if n <= 0:
        return 1
    return f(n - 1) + f(n - 2) + f(n // 2)

a = f(20)
print()
print(a)
````

Определение:

```text
0! = 1
n! = n · (n - 1)!
```

очень естественно переводится в рекурсивную функцию.
